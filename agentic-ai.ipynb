```python
# 1. Imports
from dotenv import load_dotenv
from groq import Groq
import os
import time
# 2. loading .env file 
load_dotenv()

# 3. found key
groq_key = os.getenv("GROQ_API_KEY")

# 4. making client
client = Groq(api_key=groq_key)

# 5.sending Message 
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": "Say hello in one line"}
    ]
)

# 6. Response print 
print(response.choices[0].message.content)
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
def get_page_text_anakin(url):
    try:
        anakin_key = os.getenv("ANAKIN_API_KEY")
        headers = {
            "X-API-Key": anakin_key,
            "Content-Type": "application/json"
        }
        payload = {
            "url": url,
            "formats": ["markdown"]
        }
        job = requests.post("https://api.anakin.io/v1/url-scraper", headers=headers, json=payload).json()
        job_id = job["jobId"]
        time.sleep(5)
        result = requests.get(f"https://api.anakin.io/v1/url-scraper/{job_id}", headers=headers).json()
        return result.get("markdown", "")
    except Exception as e:
        print("Anakin scraping failed:", e)
        return ""
```

# Run agent

```python
def run_agent(topic):
    results = search_web(topic, max_results=3)

    all_texts = []
    for r in results:
        text = get_page_text_anakin(r['href'])
        all_texts.append(text)

    combined_text = " ".join(all_texts)
    combined_text = combined_text[:6000]

    if len(combined_text.strip()) == 0:
        print("No content found")
    else:
        prompt = "Summarize the following content in simple points:\n\n" + combined_text
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}]
        )

        sources_text = "\n\n## Sources\n"
        for r in results:
            sources_text += f"- [{r['title']}]({r['href']})\n"

        filename = topic.replace(" ", "_") + "_summary.md"
        with open(filename, 'w', encoding='utf-8') as f:
            f.write(response.choices[0].message.content + sources_text)
        print("Done! Saved as", filename)
```

## comparing function

```python
def compare_agent(option1, option2):
    results1 = search_web(option1, max_results=2)
    results2 = search_web(option2, max_results=2)

    text1 = ""
    for r in results1:
        text1 += get_page_text_anakin(r['href'])

    text2 = ""
    for r in results2:
        text2 += get_page_text_anakin(r['href'])

    text1 = text1[:3000]
    text2 = text2[:3000]

    if len(text1.strip()) == 0 or len(text2.strip()) == 0:
        print("Not enough content found for comparison")
        return

    prompt = f"""Compare {option1} vs {option2} based on the following information.

{option1} info:
{text1}

{option2} info:
{text2}

Give clear pros and cons for each, and end with a specific recommendation on which is better and why."""

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}]
    )

    sources_text = f"\n\n## Sources\n### {option1}\n"
    for r in results1:
        sources_text += f"- [{r['title']}]({r['href']})\n"
    sources_text += f"\n### {option2}\n"
    for r in results2:
        sources_text += f"- [{r['title']}]({r['href']})\n"

    filename = f"comparison_{option1.replace(' ', '_')}_vs_{option2.replace(' ', '_')}.md"
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(response.choices[0].message.content + sources_text)

    print(f"Comparison saved to {filename}")
    print("\nPreview:\n", response.choices[0].message.content[:300], "...")
```

```python
run_agent("best productivity apps 2026")
```

```python
run_agent("climate change solutions")
```
