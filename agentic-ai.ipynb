```python
# 1. Imports
from dotenv import load_dotenv
from groq import Groq
import os

# 2. loading .env file 
load_dotenv()

# 3. foun key
api_key = os.getenv("GROQ_API_KEY")

# 4. making client
client = Groq(api_key=api_key)

# 5.sending Message 
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": "Say hello in one line"}
    ]
)

# 6. Response print 
print(response.choices[0].message.content)
```

```python
# Library install 
!pip install ddgs

# Import
from ddgs import DDGS

# Search function
def search_web(query, max_results=5):
    with DDGS() as ddgs:
        results = ddgs.text(query, max_results=max_results)
        return results

# Test 
results = search_web("agentic AI hackathons 2026")
for r in results:
    print(r['title'])
    print(r['href'])
    print(r['body'])
    print("---")
```

```python
import requests
from bs4 import BeautifulSoup

def get_page_text(url):
    try:
        res = requests.get(url)
        soup = BeautifulSoup(res.text, 'html.parser')
        all_p = soup.find_all('p')
        texts = []
        for p in all_p:
            texts.append(p.get_text(strip=True))
        return texts
    except Exception as e:
        print("Failed:", e)
        return []
```

# Summary report

```python
def run_agent(topic):
    results = search_web(topic, max_results=3)
    results = search_web(topic, max_results=3)
    
    all_texts = []
    for r in results:
        text = get_page_text(r['href'])
        all_texts.append(text)
    
    combined_text = ""
    for text_list in all_texts:
        combined_text += " ".join(text_list)
    combined_text = combined_text[:6000]  
    if len(combined_text.strip()) == 0:
        print("No content found")
    else:
        prompt = "Summarize the following content in simple points:\n\n" + combined_text
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[{"role": "user", "content": prompt}]
        )
        with open("summary.md", 'w', encoding='utf-8') as f:
            f.write(response.choices[0].message.content)
        print("Done!")
```

```python
run_agent("best productivity apps 2026")
```
